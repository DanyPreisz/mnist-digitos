# Clasificación de Dígitos Manuscritos (MNIST)

Proyecto de visión por computadora listo para **Google Colab**.

**Pregunta:** ¿podemos clasificar dígitos del 0 al 9?

**Método:** CNN simple con Keras/TensorFlow

**Stack:** TensorFlow, matplotlib

Dataset MNIST.

## 0. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
print("OK")
```

## 1. Cargar datos

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

## 2. EDA

```python
plt.figure(figsize=(8, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(y_train[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 3. Preprocesamiento

```python
X_train = X_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0
print(X_train.shape)
```

## 4. Modelo CNN

```python
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

## 5. Entrenamiento

```python
history = model.fit(X_train, y_train, epochs=3, batch_size=64, validation_split=0.1)
print("Entrenamiento completado")
```

## 6. Evaluación

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"Accuracy en test: {test_acc:.4f}")
```

```python
plt.figure(figsize=(8, 4))
plt.plot(history.history["accuracy"], label="train")
plt.plot(history.history["val_accuracy"], label="val")
plt.title("Accuracy")
plt.legend()
plt.show()
```

## 7. Predicciones de ejemplo

```python
preds = model.predict(X_test[:10])
pred_labels = np.argmax(preds, axis=1)

plt.figure(figsize=(8, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_test[i].reshape(28, 28), cmap="gray")
    plt.title(f"Pred: {pred_labels[i]}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 8. Conclusiones

- CNN simple alcanza ~98% de accuracy en MNIST.
- Dataset clásico para empezar con visión por computadora.
- Se puede extender a Fashion-MNIST o CIFAR-10.
